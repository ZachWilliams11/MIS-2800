## Part B1 — Grouping and counting (raw text)
Which product categories have the most products, and how many products are in each category (sorted from most to least)

```python
%%sql
-- Group by category and count.
SELECT category, COUNT(*) AS product_count FROM inventory_table GROUP BY category ORDER BY product_count DESC


```

## Part B2 — Fix grouping with LOWER() - individual one

```python
%%sql
-- Make category case-insensitive: group by LOWER(category) and count (ignore NULLs)
SELECT LOWER(category) AS new_category, COUNT(*) AS product_count FROM inventory_table GROUP BY new_category ORDER BY product_count DESC
```

## Part B3 — Fix spaces with TRIM() + LOWER()

```python
%%sql

-- Clean spaces and case: group by LOWER(TRIM(category)) and count (ignore NULLs).
SELECT category, COUNT(*) AS product_count FROM inventory_table GROUP BY LOWER(TRIM(category)) ORDER BY product_count DESC
```

## Part C1 — Order by category (alphabetical)

```python
%%sql
-- Show unique cleaned categories in alphabetical order.

SELECT category, COUNT(*) AS product_count FROM inventory_table GROUP BY LOWER(TRIM(category)) ORDER BY category ASC
```

## Part D1 — Case matters

```python
%%sql
-- Return rows where category = 'apple' (exact match).
SELECT category, COUNT(*) as product_count FROM inventory_table WHERE category == 'apple' GROUP BY category
```

## Part D2 — Case-insensitive match

```python
%%sql
-- Return rows that are apple regardless of case/spaces using LOWER(TRIM(category)) = 'apple'.
SELECT category, COUNT(*) as product_count FROM inventory_table WHERE LOWER(TRIM(category)) == 'apple' GROUP BY category
```

## Part D3 — Empty strings aren’t NULL

```python
%%sql
-- Show rows where category is an empty string '' (NOT NULL).
SELECT category, COUNT(*) as product_count FROM inventory_table WHERE category == '' GROUP BY category
```

## Part D4 — Find NULL categories

```python
%%sql
-- Show rows where category IS NULL.
SELECT category, COUNT(*) as product_count FROM inventory_table WHERE category IS 'None' GROUP BY category

```
